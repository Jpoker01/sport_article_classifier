# Tokenization analysis

The goal of this notebook is to pick a sensible `max_length` for the
transformer fine-tuning by inspecting how many subword tokens each
article produces under the candidate tokenizers.

Runs on the cleaned parquet, so `data_prep.ipynb` must be executed first.

```r
import pandas as pd
from transformers import AutoTokenizer

from src.config import CLEAN_DATA_PATH

df = pd.read_parquet(CLEAN_DATA_PATH)
```

### Tokenizing using robeczech-base

```r
df = pd.read_parquet(CLEAN_DATA_PATH)

tokenizer = AutoTokenizer.from_pretrained("ufal/robeczech-base")
lengths = df["text"].map(lambda t: len(tokenizer.encode(t, add_special_tokens=True)))
print(lengths.quantile([0.5, 0.9, 0.95, 0.99, 1.0]))
```

### Tokenizing using small-e-czech

```r
tokenizer = AutoTokenizer.from_pretrained("Seznam/small-e-czech")
lengths = df["text"].map(lambda t: len(tokenizer.encode(t, add_special_tokens=True)))
print(f"\n{model_name}")
print(lengths.quantile([0.5, 0.9, 0.95, 0.99, 1.0]))
```
